```python
# yoloe 11m seg에서 pf(Prompt Free) 모델
from ultralytics import YOLOE

# 1. 프롬프트 프리 모델 로드
# pf가 붙은 모델은 미리 학습된 클래스를 자동 탐지함
model = YOLOE("yoloe-11s-seg-pf.pt")

# 2. 예측 실행
# source='4'는 강사님 환경의 카메라 번호임. 본인 환경에 맞게 수정 필요
# results = model.predict(source='4', show=True)

# 2. 예측 실행 (이미지 경로 지정)
results = model.predict(
    source='/home/rokey/ros2_ws/images/fruit2.jpg',
    save=True,  # 결과를 파일로 저장
    show=True,   # 화면에도 띄움
    imgsz=640,    # 이미지 크기 제한
    device='cpu'  # [핵심] 그래픽카드 끄고 CPU로 계산
)
```

```python
# yoloe 11m seg에서 프롬프트 기반 모델(프롬프트 제공 X)
from ultralytics import YOLOE
# from ultralytics import YOLO

# 1. 프롬프트 기반 모델 로드
model = YOLOE("yoloe-11m-seg.pt")

# 2. 감지할 클래스 목록(텍스트 프롬프트) 설정
# 주석 처리되어 있어 클래스 정보가 전달되지 않음 -> 감지 실패
model.set_classes(["grapefruit", "avocado", "grapes", "orange", "mango", "banana"])



# 3. 예측 실행
# results = model.predict(source='4', show=True)

# 3. 예측 실행 (이미지 경로 지정)
results = model.predict(
    source='/home/rokey/ros2_ws/images/fruit2.jpg',
    save=True,
    show=True,
)
```
